# traj-eval quickstart

Load synthetic trajectories, run the evaluation, and look at convergence. No model weights needed.

```python
import sys
sys.path.insert(0, "../src")
from traj_eval.synthetic import generate_synthetic_trajectories
from traj_eval.metrics import evaluate_trajectories

trajs = generate_synthetic_trajectories(n_per_archetype=5)
print(f"Loaded {len(trajs)} trajectories")
print([t.id for t in trajs[:3]])
```

```python
results = evaluate_trajectories(trajs)
print(results["aggregate"])
```

```python
# Per-trajectory table (no heavy deps)
for row in results["per_trajectory"]:
    print(f"{row['id']:<20} final={row['final_prediction']} gold={row['gold_answer']} conv={row['convergence_step']} flip={row['flip_rate']:.2f} rat={row['rationalization_score']:.2f}")
```

```python
# Convergence histogram - matplotlib if available, else text counts
try:
    import matplotlib.pyplot as plt
    fracs = [r["convergence_fraction"] for r in results["per_trajectory"] if r["convergence_fraction"] is not None]
    plt.hist(fracs, bins=5)
    plt.xlabel("convergence fraction (conv_idx / n_steps)")
    plt.ylabel("count")
    plt.title("When do trajectories converge?")
    plt.show()
except ImportError:
    from collections import Counter
    fracs = [round(r["convergence_fraction"], 1) for r in results["per_trajectory"] if r["convergence_fraction"] is not None]
    print(Counter(fracs))
```
